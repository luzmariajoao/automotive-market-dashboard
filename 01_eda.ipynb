# European Car Market 2024 - Exploratory Data Analysis

Análise do mercado automóvel europeu em 2024: vendas por país, top marcas, modelos best-sellers e mix de combustível.

**Fontes:** ACEA e JATO Dynamics (via best-selling-cars.com), ano completo 2024.

```python
import sys
sys.path.append('../src')

import pandas as pd
import matplotlib.pyplot as plt
from loader import (
    load_sales_by_country,
    load_top_brands,
    load_top_models,
    load_manufacturer_groups,
)

pd.set_option('display.max_rows', 50)
```

## 1. Vendas por país

```python
countries = load_sales_by_country()
print(f"Total de mercados: {len(countries)}")
print(f"Vendas totais Europa 2024: {countries['sales_2024'].sum():,}")
countries.head(10)
```

```python
top10 = countries.head(10)
plt.figure(figsize=(10, 5))
plt.barh(top10['country'][::-1], top10['sales_2024'][::-1], color='#185FA5')
plt.title('Top 10 mercados automóveis na Europa (2024)')
plt.xlabel('Novos registos')
plt.tight_layout()
plt.show()
```

## 2. Top 10 marcas

```python
brands = load_top_brands()
brands
```

```python
plt.figure(figsize=(10, 5))
colors = ['#1D9E75' if c > 0 else '#D85A30' for c in brands['pct_change']]
plt.bar(brands['brand'], brands['pct_change'], color=colors)
plt.title('Crescimento de vendas por marca (2024 vs 2023)')
plt.ylabel('% variação')
plt.axhline(0, color='black', linewidth=0.5)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 3. Top modelos e mix de combustível

```python
models = load_top_models()
models
```

```python
fuel_counts = models['fuel_type'].str.split('/').explode().value_counts()
plt.figure(figsize=(7, 7))
plt.pie(fuel_counts, labels=fuel_counts.index, autopct='%1.0f%%', startangle=90)
plt.title('Mix de combustível nos top 10 modelos (2024)')
plt.tight_layout()
plt.show()
```
